# What is MLOPS?

-> MLOPS or Machine learning Operations

-> It is essentially DevOps for machine learning. Just like DevOps manages socftware development and deployment, MLOPS manages the end to end lifecycle of machine learning.

-> MLOPS life cycle: Data -> clean -> train -> track -> validate -> deploy -> monitor -> retrain.


# Why do we need MLOPS?

-> Automates the entire ML Workflow.

-> Reduces manual work.

-> Maintains the model accuracy

-> Monitors drift and performance

-> Makes retraining easy.


# What is Devops?

DevOps = Development + Operations

It is a culture + process, that helps software teams deliver applications faster, automatically and fewwe errors.

-> It removes gap between:

Developer(who write code)

Operations(who deploy and maintain the application).

DevOps means: Write code -> test -> build -> deploy -> monitor automatically.

# Why devops?

-> Faster software delivery

-> Automatic deployment

-> Fewer bugs

-> Easy updates

-> Continous improvement

Eg: You build a food delivery app

Without DevOps:

-> Developers write code

-> Operations team manually deploy

-> manual testing

-> slow updates.

With devops:

-> Developer pushes code to git.

-> CI/CD Pipeline automatically:

1. test the code
2. Build the app
3. Deploy it to servers.

-> Deployment issues: Moving a model from developement to production can fail due to dependencies, incompatible environments or lack of testing.

-> Monitoring issues: Once deployed, models can degrade over the time and data patterns change(called data drift)

-> Lifecycle management: It means that managing the entire journey of machine learning models start to end:

It ensures the model is:

1. Built properly
2. Tracked
3. Deployed
4. Monitored
5. Retrained
6. Retired when needed

-> It is like taking care of a model from its birth to its end.

Eg: Medical prediction

1. Data collection: collect the patients data(age,bmi,glucose level)
2. data processing: clean, encode,scale the data
3. model training: train the logistic regression and test its accuracy.
4. experiment tracking: track model, version, parameters, accuracy.
5. Model validation: accuracy, bias, overfitting, fairness. Approve the best model.
6. Deployment: Deploy the model as an api for hospital use.
7. Monitoring: track (accuracy, drop, drift)
8. Retraining: If accuracy drops(eg: data drift), retring the model automatically.
9. Retirement: If a better a model is built, retire the old model.

# process of MLOPS:

1. Designing phase: This phase is all about understanding the business problem and planning the solution. It is the blueprint stage.

-> Gathering requirements from stakeholders.

-> Defining ML use cases and prioritizing them.

-> Assessing data availability and quality.

-> Specifying functional and non functional requiremnts.

* Problem definition
* Data understanding
* Model selection
* Pipeline design
* Infrsatructure planning.

2. Model development: It is the phase where we actually build the machine learning model.

-> It includes everything from preparing data -> training the model -> evaluating and improving it.

3. Operations phase: The operations phase is the stage where the ML model is deployed, monitored and miantained in real world production, it is where the model actually runs and supports users/business.
