# **4 pillars of oops**

# 1. **Inheritance**:
 inheritance means when one class uses the properities or methods of another ... it help code resuability and build relationship between classes

**types** **of** **inheritance**:

1. **single inheritance** :  one child inherits from one parent

2. **multi-level inheritance**:  child class inherits from parent, grandchild inherits from child class

3. **multiple inheritance**: child class inherits from more than one parent class

4. **hierarichal inheritance**:multiple child classes inherits from one parent class

#  **encapsulation** :
Encapsulation is a way to protect or hide the internel details of a class from outside acess .
 it helps to control whatt data or methods can be acessed or modified from outside of the class


In encapsulation , we have three acess modifier:

1. Public:  no parameter - acessible everywhere

2. Protected: _varaible - acessible in class and in subclass

3. Private:  __varaible -acesible within the class




# 3. **polymorphism**:
it means many forms..

it allows the same function or method to behave differently depending on the object it is acting upon



# 4. **Abstraction**

abstraction means hiding unecessary details from the user and show only
essential information ...


in python, abstraction is used to hide internel implementation of a class
and expose only relevent funtion and properties

```python
# granparent -> parent -> child
# whenever we have more than one constructor , we have to call the constructors manually in the child class
class grandparent:

  def __init__(self):
    self.balance = "1,90,23,3990"

class Parent(grandparent):
  def __init__(self):
    grandparent.__init__(self) # calling the parent constructor
    self.job = "engineer"

  def display(self):
    print(self.balance , self.job)

class child(Parent):
  def __init__(self):
    Parent.__init__(self)
    self.college_degree = "B.com"

  def display(self):
    print(self.job, self.college_degree, self.balance)

obj = Parent()
obj.display()

obj1 = child()
obj1.display()
```

```python
from abc import ABC, abstractmethod
class Appliance(ABC):

  def __init__(self,name):
    self.name = name   # constructor in abstract class
  @abstractmethod
  def turn_on(self):
    print("turning on the robot using a13222")

  def details(self):
    print(f"{self.name} is an electrical appliance..")

class Washingmachine(Appliance):
  def __init__(self,name,capacity):
    super().__init__(name) # calling constructor of abstract class
    self.capacity = capacity

  def turn_on(self):
    print(f"{self.name} is not running with {self.capacity} kg load")


  def turn_off(self):
    print("turn off the machine")

w = Washingmachine("LG washing machine", 7)
w.turn_on()
w.details()
w.turn_off()
```

```python
from abc import ABC, abstractmethod

class BankAccount(ABC):
    def __init__(self, principal, years):
        self.principal = principal
        self.years = years


    @abstractmethod
    def get_interest_rate(self):
        pass


    def calculate_interest(self):
        rate = self.get_interest_rate()

        interest = (self.principal * rate * self.years) / 100
        return interest

class SavingsAccount(BankAccount):
    def get_interest_rate(self):

        return 4.5


account = SavingsAccount(principal=10000, years=3)
interest = account.calculate_interest()
print(f"Interest for {account.years} years at {account.get_interest_rate()}% rate is: {interest}")
```

```python
from abc import ABC, abstractmethod

class LibraryItem(ABC):
    @abstractmethod
    def get_description(self):
        print("This method should be overridden by subclasses.")

class Book(LibraryItem):
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def get_description(self):
        return f"Book: '{self.title}' by {self.author}"

class Magazine(LibraryItem):
    def __init__(self, title, serial_number):
        self.title = title
        self.serial_number = serial_number

    def get_description(self):
        return f"Magazine: '{self.title}', Issue #{self.serial_number}"

if __name__ == "__main__":
    book = Book("Robin", "George Orwell")
    magazine = Magazine("National Geographic", 203)

    print(book.get_description())
    print(magazine.get_description())
```

```python
#3. create a base class shape with a method area()
# create subclasses circle and rectange that implement area() differently

import math
class shape:
  def area(self):
    pass


class circle(shape):
  def __init__(self, radius):
    self.radius = radius
  def area(self):
    return math.pi*self.radius**2   #

class Rectange(shape):
  def __init__(self,width, height):
    self.width = width
    self.height = height

  def area(self):
    return self.width*self.height


c = circle(5)
c.area()
r = Rectange(4, 6)
r.area()
c = circle(5)
c.area()
```
