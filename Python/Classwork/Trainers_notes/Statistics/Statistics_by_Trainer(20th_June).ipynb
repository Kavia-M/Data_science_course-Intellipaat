**WHAT IS STATISTICS?**
- Statistics is a way of collecting, organizing, analyzing, and interpreting data to understand things better and make decisions.



**NEED FOR STATISTICS:**
- To make sense of data: understand patterns
- To make predictions
- To make better decision
- To find relationships
- To test ideas

**WHAT DOES STATISTICS DO?**
- **Collects Data** - Surveys,experiments or observations
- **Summarizes Data** - Using numbers like averages, percentages or graphs
- **Finds Patterns** - Identifies trends, relationships and predictions
- **Helps in Decision-Making** - Used in business, healthcare, sports, etc.

**TYPES OF STATISTICS:**

1. **DESCRIPTIVE STATISTICS:**
   - Summarize and describe the characteristics of a dataset.
   - Used to analyze and present data in a meaningful way.
   - Examples: Mean,Median,Standard Deviation,Correlation, etc.

2. **INFERENTIAL STATISTICS:**
   - Draws conclusions or makes predictions about a population based on a sample.
   - Used to generalize findings beyond the dataset.
   - Examples: Hypothesis Testing, Confidence Intervals, etc.

**Population**- The entire data, superset

**Sample**- subset of population, a small group from a larger dataset.

```python
#SAMPLING: process of picking data from a population for a sample
#helps in making predictions without analyzing the whole dataset

import pandas as pd

population=pd.Series(range(1,101))

s=population.sample(10)

print("Population mean", population.mean())
print("Sample mean",s.mean())
```

```python
#Random Sampling: Every data point has a equal chance of being selected
#Example: drawing names from a hat

random_sample=population.sample(10, random_state=10)
print("Random Sample\n", random_sample)
```

```python
#Stratified Sampling: Divide population into groups(strata) and sample from each group
#Example: Survey based on stream(Science/Commerce,Arts)

import pandas as pd
from sklearn.model_selection import train_test_split

df=pd.DataFrame({
    'score':range(1,101),
    'grade':['A']*30+['B']*40+['C']*30
})

#train_test_split returns 2 outputs, 1: 10% that goes into strat_sample, 2:90% that goes into _ because we dont need it
strat_sample,_=train_test_split(df,stratify=df['grade'],test_size=0.9)

print(strat_sample['grade'].value_counts())
#value_counts() function gives count for each kind of value in grade
```

```python
#Systematic Sampling: The values are selection for every kth value

import pandas as pd

data={
    'ID':range(1,101),
    'Value':range(100,200)
    }

df=pd.DataFrame(data)

sample_size=10 #how many samples do i want
population_size=len(df) #Total rows in the dataset

k=population_size//sample_size #sampling interval

start=3
systematic_sample=df.iloc[start::k]

print(systematic_sample)
```

**Central Tendencies**
- Mean(average): Sum of all values divided by the count
- Median(middle value)
- Mode(most frequent value)

```python
import numpy as np
from statistics import mean,median,mode

data=[10,20,20,25,30,30,30,35,40,100]

avg=np.sum(data)/len(data)
print(avg)
print("Mean", mean(data))

sortedData=sorted(data)
mid=len(sortedData)//2
print(mid)
print(sortedData[mid])
print("Median", median(data))


freq={}

for i in data:
  freq[i]=freq.get(i,0)+1

max_count=max(freq.values())

for number in freq:
  if freq[number]==max_count:
    print("Mode",number)
    break

print("Mode(function)",mode(data))
```

**Measures of Variation**
- tells  us how spread out the data is.

- **Range**
  - Max-Min, gives the total spread

- **Percentiles**
  - Shows value below which a certain percentage of data falls

- **Quartiles**(special percentiles)
  - 25,50,75

- **Variance**
  - how far each number is from the mean, on average.

- **Standard Deviation**
   - Square root of variance, tells average distance from the mean

- **Correlation**
   - tells how two variables move together
   


```python
print(data)
```

```python
range_val=max(data)-min(data)
print("Range", range_val)
```

```python
#Percentile
print("25th Percentile", np.percentile(data,25))
print("50th Percentile", np.percentile(data,50))
print("90th Percentile", np.percentile(data,90))
```

```python
#Quartile
print("25th Percentile", np.percentile(data,25))
print("50th Percentile", np.percentile(data,50))
print("75th Percentile", np.percentile(data,75))
```

```python
#Variance
variance=np.var(data,ddof=1)
print("Variance",variance)
```
