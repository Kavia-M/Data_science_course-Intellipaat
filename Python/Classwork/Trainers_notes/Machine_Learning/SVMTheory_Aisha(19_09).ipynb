#1. What is SVM?
- Support Vector Machine is a supervised learning algorithm.
- It can be used for both regression and classification tasks.
    - Its **main goal** is to find the best boundary(hyperplane) that separates data into different classes.
    - For regression, it tries to fit a line/plane within a margin that minimizes error.

**KERNEL TRICK**: changing the dimension and using a hyperplane to separate different data points(classes).

#HARD Margin vs SOFT Margin

- **Hard Margin**
    - Strictly separates the classes without errors.
    - Works when data is perfectly linearly separable.
    - Problem: Very sensitive to outliers

- **Soft Margin**
    - Allows some misclassification.
    - Balance between maximizing margin and minimizing error.
    - More practical in real world data.



#Linear vs Polynomial

linearpolykernel.png


**Linear Kernel**
- No transformation
- Works when data is linearly separable

**Polynomial Kernel**
- Transforms features into polynomial space
- Helps separate non-linear data with curved boundaries.

**Real Life Use Cases**
- Image Classification
- Text Classification
