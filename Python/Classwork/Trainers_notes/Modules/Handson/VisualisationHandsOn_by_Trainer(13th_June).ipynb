1. - Create scatter plots for the following data, make sure all the plots appear in the same plane with x-labels and y-labels evenly spaced.

- Create scatter plots between (mpg-hp, mpg-weight), and (hp-mpg, hp-time-to-60) with respect to each brand.

```python
fig=plt.subplots(2,2, figsize=(20,10))

plt.subplot(2,2,1) #r,c,g
sns.scatterplot(data=df, x='mpg', y='hp', hue='brand')
plt.title("Mpg vs Hp")
plt.xlabel("Mpg")
plt.ylabel("Hp")

plt.subplot(2,2,2) #r,c,g
sns.scatterplot(data=df, x='mpg', y='weightlbs',hue='brand')
plt.title("Mpg vs Weight")
plt.xlabel("Mpg")
plt.ylabel("Weight")

plt.subplot(2,2,3) #r,c,g
sns.scatterplot(data=df, x='hp', y='mpg',hue='brand')
plt.title("Hp vs Mpg")
plt.ylabel("Mpg")
plt.xlabel("Hp")

plt.subplot(2,2,4) #r,c,g
sns.scatterplot(data=df, x='hp', y='time-to-60', hue='brand')
plt.title("hp vs time-to-60")
plt.xlabel("hp")
plt.ylabel("time-to-60")

plt.show()
```

```python
fig=plt.subplots(1,1, figsize=(10,90))

plt.subplot(1,1,1) #r,c,g
sns.scatterplot(data=df, x='mpg', y='weightlbs',hue='brand')
plt.title("Mpg vs Weight")
plt.xlabel("Mpg")
plt.ylabel("Weight")
plt.show()
```

2. - Create bar plots for the following, make sure all the plots appear in the same plane with x-labels and y-labels evenly spaced.

- Create a bar plot that shows the visual representation of hp, mpg, weight, time-to-60 with respect to the number of cylinders for each of the brands.


```python
fig=plt.subplots(2,2, figsize=(20,10))

plt.subplot(2,2,1) #r,c,g
sns.barplot(data=df, y='mpg', x='cylinders', hue='brand')
plt.title("Mpg vs Cylinder")
plt.xlabel("Cylinder")
plt.ylabel("Mpg")

plt.subplot(2,2,2) #r,c,g
sns.barplot(data=df, y='hp', x='cylinders',hue='brand')
plt.title("Hp vs cylinders")
plt.xlabel("Hp")
plt.ylabel("cylinders")

plt.subplot(2,2,3) #r,c,g
sns.barplot(data=df, y='weightlbs', x='cylinders',hue='brand')
plt.title("Weight vs cylinders")
plt.ylabel("Weight")
plt.xlabel("cylinders")

plt.subplot(2,2,4) #r,c,g
sns.barplot(data=df, x='time-to-60', y='cylinders', hue='brand')
plt.title("time-to-60 vs cylinders")
plt.ylabel("hp")
plt.xlabel("time-to-60")

plt.show()
```
