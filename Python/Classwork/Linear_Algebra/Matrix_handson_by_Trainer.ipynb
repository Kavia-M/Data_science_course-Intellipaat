# Property 2: (A + B).T= A.T + B.T

```python
A=np.array([[12,14],[10,8]])
B=np.array([[1,4],[20,16]])

x=(A+B).T
print(x)
y=A.T+B.T
print(y)
```

# Property:3 (AxB).T=B.TxA.T

```python
A=np.array([[12,14],[10,8]])
B=np.array([[1,4],[20,16]])

x=np.dot(A,B).T
print(x)

y=B.T.dot(A.T)
print(y)

if np.array_equal(x,y):
  print('True')
else:
  print('false')
```
