```python
import numpy as np
from statistics import mean,median,mode
data=[10,20,20,25,30,30,30,35,40,100]
avg=np.sum(data)/len(data)
print(avg)
print("Mean", mean(data))
sortedData=sorted(data)
mid=len(sortedData)//2
print(mid)
print(sortedData[mid])
print("Median", median(data))
```

```python
dict1 = {}

for i in data:
  dict1[i] = dict1.get(i, 0) + 1

max(dict1, key=dict1.get)
```

```python
# others code

mode_data = max((data.count(item), item) for item in set(data))[1]
print("Mode of data: ",mode_data)
```
