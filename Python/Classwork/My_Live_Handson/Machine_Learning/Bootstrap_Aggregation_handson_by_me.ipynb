```python
#Outliers: only Senior Citizen, Tenure, Monthly Charges columns, out of which only Senior citizen has outliers WITH NO BOX PLOT
```

```python
#Label Encoding
from sklearn.preprocessing import LabelEncoder

LE=LabelEncoder()

for col in churn_data.columns:
  if(churn_data[col].dtype=='object'):
    churn_data[col]=LE.fit_transform(churn_data[col])
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X=churn_data.iloc[:,:-1]
y=churn_data['Churn']
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
X_scaled
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
#stratify = y, to equally divide values of target column between training and testing data
X_train, X_test, y_train, y_test =train_test_split(X_scaled,y,test_size=0.2,random_state=0, stratify=y)
```

```python
y_train.value_counts()
```

```python
from random import Random
from sklearn.ensemble import RandomForestClassifier
```

```python
classifier1 = RandomForestClassifier(n_estimators= 100)
```

```python
classifier1.fit(X_train, y_train)
```

```python
y_pred = classifier1.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_pred, y_test)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test, y_pred)
```

```python
from sklearn.metrics import recall_score
```

```python
sensitivity = recall_score(y_test, y_pred)
```

SMOTE (Synthetic Minority Over-sampling Technique) is a method used in machine learning to address class imbalance in datasets, where one class (the minority class) has significantly fewer samples than another (the majority class). Instead of simply duplicating existing minority class samples, SMOTE generates new, synthetic samples.

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

```python
#GOAL: help the model learn patterns of the minority class better
from imblearn.over_sampling import SMOTE

smote=SMOTE(random_state=0)

X_resample,y_resample=smote.fit_resample(X_train,y_train)

classifier1.fit(X_resample, y_resample)

y_pred_smote=classifier1.predict(X_test)
```

```python
y_resample.value_counts()
```
