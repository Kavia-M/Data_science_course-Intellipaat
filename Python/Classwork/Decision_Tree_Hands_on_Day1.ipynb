 **Decision tree is a non-parametric supervised learning algorithm used for both classification and regression tasks. It is a hierarchical, tree-like structure that consists of a root node, branches, internal nodes, and leaf nodes. The root node represents the entire dataset and the initial decision to be made. Internal nodes represent decisions or tests on attributes, branches represent the outcome of these decisions, and leaf nodes represent the final outcomes or predictions.**

```python
# step 5:

# model building

from sklearn.model_selection import train_test_split

from sklearn.metrics import accuracy_score, confusion_matrix

from sklearn.tree import DecisionTreeClassifier

```

```python
x = df.drop(columns =['target'])
y = df['target']
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 45)
```

```python
model = DecisionTreeClassifier()
```

```python
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
compare = pd.DataFrame({'Actual': y_test,
                        "Predicted": y_pred})

print(compare)
```

```python
accuracy_score(y_pred, y_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
from sklearn import tree

tree.plot_tree(model , fontsize = 7)
```

```python
plt.figure(figsize =(28, 14))
tree.plot_tree(model,
               feature_names = x_train.columns,
               class_names = ['no_disease', 'disease'],
               filled = True,
               fontsize  = 15)

plt.title('decision tree visualization', fontsize =14, fontweight ='bold')
plt.show()
```
