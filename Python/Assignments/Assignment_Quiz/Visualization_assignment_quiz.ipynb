👉 A boxplot shows:

Median (2nd quartile, Q2) → the line inside the box.

1st quartile (Q1, 25th percentile) → lower edge of the box.

3rd quartile (Q3, 75th percentile) → upper edge of the box.

Interquartile Range (IQR = Q3 − Q1) → the box height.

Minimum & Maximum (whiskers) → range of most data.

Outliers → dots outside whiskers.

So, a boxplot gives all quartiles + spread + outliers → not just Q2 or Q3.

4) In a scatter plot, what does the 'hue' parameter represent?
- The color of data points based on a categorical variable
- The shape of data points
- The size of data points
- The transparency of data points

Answer- The color of data points based on a categorical variable

```python
sns.scatterplot(x="sqft_living", y="price", hue="waterfront", style="grade", size="bedrooms", data=df)
plt.show()
```

Here:

hue="waterfront" → 0 or 1 shown in different colors.

style="grade" → grade of house affect style

size="bedrooms" → number of bedroomns changes point size.

```python
import matplotlib.pyplot as plt
import numpy as np

# Sample data
x = np.random.rand(20)
y = np.random.rand(20)
categories = np.random.choice(['A','B'], size=20)   # categorical variable
sizes = np.random.randint(50, 300, size=20)         # different sizes

# Map categories to colors and shapes
color_map = {'A': 'red', 'B': 'blue'}
marker_map = {'A': 'o', 'B': 's'}   # circle for A, square for B

for cat in np.unique(categories):
    plt.scatter(
        x[categories == cat],
        y[categories == cat],
        c=color_map[cat],           # color (like hue)
        marker=marker_map[cat],     # shape
        s=sizes[categories == cat], # size
        alpha=0.6,                  # transparency
        label=f"Category {cat}"
    )

plt.legend()
plt.title("Custom Scatter Plot with Color, Shape, Size, Transparency")
plt.show()
```

With Matplotlib’s plt.scatter(), you don’t get a direct hue like in Seaborn, but you can still control:

Hue / Color → c=

Shape (marker) → marker=

Size → s=

Transparency → alpha=

What this does:

cat means category

c=color_map[cat] → different colors based on category (hue equivalent).

marker=marker_map[cat] → different shapes.

s= → bubble sizes.

alpha=0.6 → makes points semi-transparent.

⚡ This is the Matplotlib way. In Seaborn, you get all this with one function (sns.scatterplot(hue=, style=, size=, alpha=)).

4. Answer - The color of data points based on a categorical variable

6) How to create a subplot in a figure with three plots side by side and two rows of subplots?
- ax1 = fig.add_subplot(1,2,3)
- ax1 = fig.add_subplot(3,2,1)
- ax1 = fig.add_subplot(2,3,1)
- ax1 = fig.add_subplot(3*2)

```python
import matplotlib.pyplot as plt

# Choose some interesting columns
features = ["sqft_living", "bedrooms", "bathrooms", "floors", "grade", "yr_built"]

fig = plt.figure(figsize=(15, 8))

for i, col in enumerate(features, 1):
    ax = fig.add_subplot(2, 3, i)   # 2 rows × 3 cols
    ax.scatter(df[col], df["price"], alpha=0.3)
    ax.set_xlabel(col)
    ax.set_ylabel("price")

plt.tight_layout()
plt.show()
```

```python
df.info()
```

🔎 What happens here?

fig.add_subplot(2,3,i) → makes a 2×3 grid (6 subplots).

Each subplot shows scatter plot of feature vs price.

Example:

Subplot 1 → sqft_living vs price

Subplot 2 → bedrooms vs price

… and so on.

Answer is using 1st position

(because 2 rows, 3 columns, first position).
