11) Write a function to perform the following operation: Increase the grade of the house by 1 if the sqft of the house is greater than 0 and less than equal to 400. If the total sqft of the house is greater than 400, increase the grade of the house by 2. After creating the above function, which syntax will be using when using apply function to run the above created function in the dataset?
- data.app(data)
- None of the above
- data.apply(your_function_name(arg1))
- data.apply_fun(function_name())

```python
def update_grade(row):
    if 0 < row['sqft_living'] <= 400:
        row['grade'] += 1
    elif row['sqft_living'] > 400:
        row['grade'] += 2
    return row
```

```python
df.apply(update_grade, axis=1)
```

Merge definition of parameters

eft : DataFrame or named Series
right : DataFrame or named Series
    Object to merge with.
how : {'left', 'right', 'outer', 'inner', 'cross'}, default 'inner'
    Type of merge to be performed.


left: use only keys from left frame, similar to a SQL left outer join;
preserve key order.

right: use only keys from right frame, similar to a SQL right outer join;
preserve key order.

outer: use union of keys from both frames, similar to a SQL full outer
join; sort keys lexicographically.

inner: use intersection of keys from both frames, similar to a SQL inner
join; preserve the order of the left keys.

cross: creates the cartesian product from both frames, preserves the order
of the left keys.
on : label or list
