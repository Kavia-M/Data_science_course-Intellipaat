Let's solve this **carefully step by step**.

---

### Given:

* Population mean $\mu = 100$
* Population standard deviation $\sigma = 15$
* Sample:

$$
[88, 92, 84, 64, 96, 97, 67, 97, 96, 99, 115, 119, 119, 129, 130, 142, 102, 123, 134, 105]
$$

* Sample size $n = 20$

---

### Step 1: Compute the sample mean

$$
\bar{x} = \frac{\text{sum of sample}}{n}
$$

Sum of sample = $88+92+84+64+96+97+67+97+96+99+115+119+119+129+130+142+102+123+134+105$

Calculating step by step:

* First 10: $88+92=180, 180+84=264, 264+64=328, 328+96=424, 424+97=521, 521+67=588, 588+97=685, 685+96=781, 781+99=880$
* Next 10: $880+115=995, 995+119=1114, 1114+119=1233, 1233+129=1362, 1362+130=1492, 1492+142=1634, 1634+102=1736, 1736+123=1859, 1859+134=1993, 1993+105=2098$

$$
\bar{x} = \frac{2098}{20} = 104.9
$$

---

### Step 2: Compute the z-statistic

Formula:

$$
z = \frac{\bar{x} - \mu}{\sigma / \sqrt{n}}
$$

$$
z = \frac{104.9 - 100}{15 / \sqrt{20}} = \frac{4.9}{15/4.4721} = \frac{4.9}{3.354} \approx 1.46
$$

---

### Step 3: Compute the p-value

* **Two-tailed test**: $p = 2 \times P(Z > |z|)$
* For $z = 1.46$, $P(Z > 1.46) \approx 0.072$

$$
p \approx 2 \times 0.072 = 0.144
$$

```python
import numpy as np
from scipy.stats import norm

# Given data
sample7 = np.array([88, 92, 84, 64, 96, 97, 67, 97, 96, 99,115, 119, 119, 129, 130, 142, 102, 123, 134, 105])
mu = 100          # Population mean
sigma = 15        # Population standard deviation
n = len(sample7)   # Sample size

# Step 1: Sample mean
x_bar = np.mean(sample7)

# Step 2: Standard error (using population standard deviation)
SE = sigma / np.sqrt(n)

# Step 3: Z-score
z = (x_bar - mu) / SE

# Step 4: Two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Sample mean:", x_bar)
print("Z-score:", z)
print("P-value:", p_value)
```

```python
import numpy as np
from scipy.stats import norm

# Given data
sample7 = np.array([88, 92, 84, 64, 96, 97, 67, 97, 96, 99,115, 119, 119, 129, 130, 142, 102, 123, 134, 105])
mu = 100          # Population mean
sigma = 15        # Population standard deviation
n = len(sample7)   # Sample size

# Step 1: Sample mean
x_bar = np.mean(sample7)

# Step 2: Standard error (using sample standard deviation)
SE = np.std(sample7, ddof=1) / np.sqrt(n)

# Step 3: Z-score
z = (x_bar - mu) / SE

# Step 4: Two-tailed p-value
p_value = 2 * (1 - norm.cdf(abs(z)))

print("Sample mean:", x_bar)
print("Z-score:", z)
print("P-value:", p_value)
```

9) A card is drawn at random from a standard deck of cards. Recall that there are 13 hearts, 13 diamonds, 13 spades and 13 clubs in a standard deck of cards.

• Let H be the event that a heart is drawn,

• Let R be the event that a red card is drawn and

• Let F be the event that a face card is drawn, where the face cards are the kings, queens and jacks.

a. If I draw a card at random from the deck of 52, what is P(H)?

b. If I draw a card at random from the deck of 52, what is P(F)?

c. If I draw a card at random, and without showing you the card, I tell you that the card is red, then what are the chances that it is a face card?
- 0.25, 0.23, 0.23
- None of the above
- 0.50 , 0.25, 0.25
- 0.50, 0.23, 0.23

Let's solve this **step by step**.

---

### Given:

* Standard deck: 52 cards
* Suits: 13 hearts (♥), 13 diamonds (♦), 13 spades (♠), 13 clubs (♣)
* Face cards: J, Q, K in each suit → 12 total face cards (3 per suit × 4 suits)
* Red cards: Hearts and Diamonds → 26 cards

Events:

* H = heart drawn
* R = red card drawn
* F = face card drawn

---

### a. $P(H)$

$$
P(H) = \frac{\text{# of hearts}}{\text{total cards}} = \frac{13}{52} = 0.25
$$

---

### b. $P(F)$

$$
P(F) = \frac{\text{# of face cards}}{\text{total cards}} = \frac{12}{52} \approx 0.23
$$

---

### c. $P(F \mid R)$

* Formula for conditional probability:

$$
P(F \mid R) = \frac{P(F \cap R)}{P(R)}
$$

* Red face cards: J, Q, K of hearts and diamonds → 6 cards

$$
P(F \cap R) = \frac{6}{52} \quad\text{and}\quad P(R) = \frac{26}{52} = 0.5
$$

$$
P(F \mid R) = \frac{6/52}{26/52} = \frac{6}{26} \approx 0.23
$$

---

### ✅ Summary of probabilities:

$$
P(H) = 0.25, \quad P(F) \approx 0.23, \quad P(F \mid R) \approx 0.23
$$

---

### ✅ Correct answer:

**0.25, 0.23, 0.23**

---


```python
# Probabilities for a standard deck
total_cards = 52
hearts = 13
diamonds = 13
spades = 13
clubs = 13
face_cards_per_suit = 3  # J, Q, K

# a) Probability of drawing a heart
P_H = hearts / total_cards

# b) Probability of drawing a face card
total_face_cards = 4 * face_cards_per_suit
P_F = total_face_cards / total_cards

# c) Probability of face card given it's red
red_cards = hearts + diamonds
red_face_cards = 2 * face_cards_per_suit  # Hearts + Diamonds
P_F_given_R = red_face_cards / red_cards

print("P(H) =", round(P_H, 2))
print("P(F) =", round(P_F, 2))
print("P(F | R) =", round(P_F_given_R, 2))
```

10) A box contains three coins: two regular coins and one fake two-headed coin (P(H)=1). a. You pick a coin at random and toss it. What is the probability that it lands heads up? b. You pick a coin at random and toss it, and get heads. What is the probability that it is the two-headed coin?
- 2/3, 1/5
- 2/5, 1/3
- 4/6 ,1/2
- None of the above

Let's carefully solve this **step by step** using **basic probability** and **Bayes’ theorem**.

---

### Given:

* 3 coins:

  1. Coin A: regular → P(H) = 0.5
  2. Coin B: regular → P(H) = 0.5
  3. Coin C: two-headed → P(H) = 1

* Pick a coin **at random** → probability of choosing any coin = $1/3$

---

### a. Probability the toss lands heads

Use **law of total probability**:

$$
P(H) = P(\text{pick A})P(H|A) + P(\text{pick B})P(H|B) + P(\text{pick C})P(H|C)
$$

$$
P(H) = \frac{1}{3} \cdot 0.5 + \frac{1}{3} \cdot 0.5 + \frac{1}{3} \cdot 1
$$

$$
P(H) = \frac{0.5 + 0.5 + 1}{3} = \frac{2}{3}
$$

✅ So probability of heads = **2/3**

---

### b. Probability that it is the two-headed coin **given heads**

Use **Bayes’ theorem**:

$$
P(\text{two-headed} \mid H) = \frac{P(H \mid \text{two-headed}) \cdot P(\text{two-headed})}{P(H)}
$$

* $P(H \mid \text{two-headed}) = 1$
* $P(\text{two-headed}) = 1/3$
* $P(H) = 2/3$ (from part a)

$$
P(\text{two-headed} \mid H) = \frac{1 \cdot \frac{1}{3}}{\frac{2}{3}} = \frac{1}{2}
$$

✅ So probability it’s the two-headed coin = **1/2**

---

### ✅ Answer:

**2/3, 1/2**

---

```python
import random

# Probability of getting head  in each coin
# Coins: two regular (0.5) and one two-headed (1)
coins = [0.5, 0.5, 1]

trials = 100000  # number of simulations
heads_count = 0
two_headed_given_heads = 0

for _ in range(trials):
    # Pick a coin at random
    coin = random.choice(coins)
    # Toss the coin
    toss = random.random() < coin  # True if heads
    if toss:
        heads_count += 1
        # Check if it was the two-headed coin
        if coin == 1:
            two_headed_given_heads += 1

# a) Probability of heads
prob_heads = heads_count / trials

# b) Probability it's two-headed given heads
prob_two_headed_given_heads = two_headed_given_heads / heads_count

print("Simulated P(H) =", round(prob_heads, 2))
print("Simulated P(Two-headed | H) =", round(prob_two_headed_given_heads, 2))
```

18) Find t-statistic for the following samples given that the mean of the distribution is 8. Sample: [5 3 3 5 5 6 6 7 3 8]
- -5.3045
- 0.0005
- -0.0026
- 3.1623

We are asked to compute the **t-statistic** for a sample, given population mean $\mu = 8$ and the sample:

$$
[5, 3, 3, 5, 5, 6, 6, 7, 3, 8]
$$

The **t-statistic formula** for a single sample is:

$$
t = \frac{\bar{x} - \mu}{s / \sqrt{n}}
$$

Where:

* $\bar{x}$ = sample mean
* $\mu$ = population mean
* $s$ = sample standard deviation
* $n$ = sample size

---

### Step 1: Sample mean $\bar{x}$

$$
\bar{x} = \frac{5 + 3 + 3 + 5 + 5 + 6 + 6 + 7 + 3 + 8}{10}
= \frac{51}{10} = 5.1
$$

---

### Step 2: Sample standard deviation $s$

Sample standard deviation formula:

$$
s = \sqrt{\frac{\sum (x_i - \bar{x})^2}{n-1}}
$$

Compute squared deviations:

* (5 − 5.1)² = 0.01
* (3 − 5.1)² = 4.41
* (3 − 5.1)² = 4.41
* (5 − 5.1)² = 0.01
* (5 − 5.1)² = 0.01
* (6 − 5.1)² = 0.81
* (6 − 5.1)² = 0.81
* (7 − 5.1)² = 3.61
* (3 − 5.1)² = 4.41
* (8 − 5.1)² = 8.41

Sum of squared deviations:

$$
0.01 + 4.41 + 4.41 + 0.01 + 0.01 + 0.81 + 0.81 + 3.61 + 4.41 + 8.41 = 26.9
$$

Divide by $n-1 = 9$:

$$
\frac{26.9}{9} \approx 2.9889
$$

Take square root:

$$
s = \sqrt{2.9889} \approx 1.728
$$

---

### Step 3: Compute t-statistic

$$
t = \frac{\bar{x} - \mu}{s / \sqrt{n}} = \frac{5.1 - 8}{1.728 / \sqrt{10}}
$$

$$
5.1 - 8 = -2.9
$$

$$
s / \sqrt{n} = 1.728 / \sqrt{10} \approx 1.728 / 3.162 \approx 0.5463
$$

$$
t = -2.9 / 0.5463 \approx -5.3045
$$

---

✅ **Answer:** `-5.3045`


```python
import numpy as np
from scipy import stats

# Sample data
sample = np.array([5, 3, 3, 5, 5, 6, 6, 7, 3, 8])

# Population mean
mu = 8

# Calculate sample mean and standard deviation
x_bar = np.mean(sample)
s = np.std(sample, ddof=1)  # ddof=1 for sample standard deviation
n = len(sample)

# Calculate t-statistic manually
t_stat = (x_bar - mu) / (s / np.sqrt(n))
print("t-statistic:", t_stat)

# Or using scipy built-in one-sample t-test
t_stat_scipy, p_value = stats.ttest_1samp(sample, mu)
print("t-statistic (scipy):", t_stat_scipy)
```

19) There are two machines for spray coating at desired thickness. Their variability should be similar if they need to be used for the same applications. To test this, various samples were collected from both machines and the sample covariance of machine 1 and 2 are 1.84 and 6.36 respectively. Comment on whether these machines can be used for the same application if the machines were truly built to have the same variance. Perform the test at 95% significance given n1 = 24 n2 = 25
- Accept null hypothesis, variance of both the machines are different
- Reject null hypothesis, variance of both the machines are same
- Accept null hypothesis, variance of both the machines are same
- Reject null hypothesis, variance of both the machines are different

This is a **variance comparison problem**.
---

### Step 1: Identify the test

We are comparing **two sample variances** to check if the population variances are equal.

* Null hypothesis ($H_0$): $\sigma_1^2 = \sigma_2^2$
* Alternative hypothesis ($H_1$): $\sigma_1^2 \neq \sigma_2^2$

We use the **F-test for equality of variances**:

$$
F = \frac{s_1^2}{s_2^2} \quad \text{(larger variance on top)}
$$

Where:

* $s_1^2 = 6.36$ (larger variance)
* $s_2^2 = 1.84$ (smaller variance)
* $n_1 = 25, n_2 = 24$

---

### Step 2: Compute F-statistic

$$
F = \frac{6.36}{1.84} \approx 3.4565
$$

---

### Step 3: Determine critical F-value

Degrees of freedom:

* $df_1 = n_1 - 1 = 25 - 1 = 24$
* $df_2 = n_2 - 1 = 24 - 1 = 23$

For a **two-tailed test at 95% significance**, the critical values are:

$$
F_{lower} = \frac{1}{F_{0.975, 23, 24}}, \quad F_{upper} = F_{0.975, 24, 23}
$$

From standard F-tables (approximate):  (here 0.975 is used since it is 2 tailed, so 5% / 2 is 2.5% ..... 100% - 2.5% is 97.5% that is 0.975

* $F_{0.975, 24, 23} \approx 2.51$
* $F_{0.025, 24, 23} \approx 1 / 2.51 \approx 0.398$

---

### Step 4: Decision

* Computed $F = 3.4565$
* Upper critical $F = 2.51$

Since $F > F_{upper}$, **we reject the null hypothesis**.

✅ **Conclusion:**

> **Reject null hypothesis, variance of both the machines are different**

---


```python
import scipy.stats as stats

# Sample variances and sample sizes
s1_sq = 6.36  # variance of machine 2 (larger)
s2_sq = 1.84  # variance of machine 1 (smaller)
n1 = 25
n2 = 24

# Compute F-statistic
F = s1_sq / s2_sq
df1 = n1 - 1
df2 = n2 - 1

# Two-tailed p-value
p_value = 2 * min(stats.f.cdf(F, df1, df2), 1 - stats.f.cdf(F, df1, df2))

print("F-statistic:", F)
print("p-value:", p_value)

# Decision at 95% significance
if p_value < 0.05:
    print("Reject null hypothesis: variances are different")
else:
    print("Accept null hypothesis: variances are the same")
```

25)  The outcomes of an experiment are {e, f, g, h} with probabilities 0.39, 0.24, 0.77, and 0.88, respectively. Let E and F denote the events {e, f} and {f, h}, respectively. Compute  (a) P(E∩F) and (b) P(E ∪ F).
- 0.24,0.63
- 1.51,0.24
- 1.12,1.51
- 0.24,1.51

* An **event** is a *set* of outcomes.

* $E = \{e, f\}$ means event $E$ occurs if the outcome is either $e$ **or** $f$.
  So, yes → $P(E) = P(e) + P(f)$.

* Similarly, $F = \{f, h\}$.

Now:

1. **Intersection $E \cap F$** = outcomes common to both → $\{f\}$.

   $$
   P(E \cap F) = P(f) = 0.24
   $$

2. **Union $E \cup F$** = outcomes in either $E$ or $F$ → $\{e, f, h\}$.

   $$
   P(E \cup F) = P(e) + P(f) + P(h) = 0.39 + 0.24 + 0.88 = 1.51
   $$

👉 Normally, probabilities should add up to ≤ 1, but here the problem statement gave inconsistent values (since 0.77 for $g$ and 0.88 for $h$ already break the rule).


```python
# Define the sample space probabilities
probs = {
    'e': 0.39,
    'f': 0.24,
    'g': 0.77,
    'h': 0.88
}

# Define the events
E = {'e', 'f'}
F = {'f', 'h'}

# Intersection (common outcomes)
E_inter_F = E.intersection(F)    # set operation
P_intersection = sum(probs[x] for x in E_inter_F)

# Union (all outcomes in E or F)
E_union_F = E.union(F)     # set operation
P_union = sum(probs[x] for x in E_union_F)

print("P(E ∩ F) =", P_intersection)
print("P(E ∪ F) =", P_union)
```

25. Answer - 0.24,1.51
