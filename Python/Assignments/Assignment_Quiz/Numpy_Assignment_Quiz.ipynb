11) For the given python code that implements bubble sort, what will be the output for the given numpy array. def bub_sort(array): for i in range(0, len(array)): for j in range(0, len(array) - i - 1): if array[j] < array[j + 1]: temp = array[j] array[j] = array[j + 1] array[j+1] = temp return array my_array = numpy.array([20,14,25,16,45,60,12,9])
- array([60, 45, 20, 25, 16, 14, 12, 9])
- array([60, 45, 25, 20, 16, 14, 12, 9])
- array([ 9, 12, 14, 16, 20, 25, 45, 60])
- array([ 9, 12, 16, 14, 20, 25, 45, 60])

````python
def bub_sort(array):
  for i in range(0, len(array)):
    for j in range(0, len(array) - i - 1):
      if array[j] < array[j + 1]:
        temp = array[j]
        array[j] = array[j + 1]
        array[j+1] = temp
  return array

my_array = np.array([20,14,25,16,45,60,12,9])
````

````python
bub_sort(my_array)
````

Perfect! Let’s do a **step-by-step table** for your descending bubble sort to see **how each pass moves the largest numbers forward and the smallest ends up last**.

---

### **Initial array**

```
Index: 0   1   2   3   4   5   6   7
Array: 20  14  25  16  45  60  12  9
```

---

### **Pass 1 (i = 0)**

Compare pairs, swap if **left < right**:

| j | Compare   | Swap? | Array after swap          |
| - | --------- | ----- | ------------------------- |
| 0 | 20 < 14 ? | No    | \[20 14 25 16 45 60 12 9] |
| 1 | 14 < 25 ? | Yes   | \[20 25 14 16 45 60 12 9] |
| 2 | 14 < 16 ? | Yes   | \[20 25 16 14 45 60 12 9] |
| 3 | 14 < 45 ? | Yes   | \[20 25 16 45 14 60 12 9] |
| 4 | 14 < 60 ? | Yes   | \[20 25 16 45 60 14 12 9] |
| 5 | 14 < 12 ? | No    | \[20 25 16 45 60 14 12 9] |
| 6 | 12 < 9 ?  | No    | \[20 25 16 45 60 14 12 9] |

**End of Pass 1:**

* Largest element `60` is near front (index 4 moved forward).
* Smallest element `9` starts moving toward the end.

---

### **Pass 2 (i = 1)**

Now last element is already in correct position for this pass:

* Array before pass: `[25 20 16 45 60 14 12 9]` (after previous swaps reorganized)

* After all swaps: `45, 60 already in place` → the next largest moves to front

* End of Pass 2: largest numbers increasingly settle at the **front**, smaller numbers drift to the **end**.

---

### **Pass 3… Pass 7**

* Repeat the same logic:

  * Largest remaining number moves forward
  * Last element gradually becomes **the smallest**
* By **final pass**, array fully sorted in **descending order**:

```
[60, 45, 25, 20, 16, 14, 12, 9]
```

---

### ✅ Key observations

1. **Largest numbers** “bubble up” toward **front (index 0)** because of `<` comparison.
2. **Smallest number `9`** sinks step by step to the **last index**.
3. The algorithm does **len(array) passes**, each time ignoring the already sorted elements at the **start (descending)**.
