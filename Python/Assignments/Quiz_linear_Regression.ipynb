14) Build a Linear regression Model to predict Selling price of the Vehicle based upon the Year , Present_Price,  Kms_Driven  ,Fuel_Type , Seller_Type  Transmission , Owner. Compute the R2 Score  for the same.

```python
df1 = df_carr_no_duplicates.drop(columns=['name'])
```

```python
df_carr_no_duplicates
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
df1
```

```python
for i in df1.columns:
  if (df1[i].dtype != 'object'):
    plt.boxplot(df1[i])
    plt.title(i)
    plt.show()
```

```python
for i in df1.columns:
  if df1[i].dtype != 'object':
    q1 = df1[i].quantile(0.25)
    q3 = df1[i].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5*iqr
    upper_limit = q3 + 1.5*iqr
    df1 = df1[(df1[i] >= lower_limit) & (df1[i] <= upper_limit)]
```

```python
encoders = {}
mappings = {}
for i in df1.columns:
  if df1[i].dtype == 'object':
    le = LabelEncoder() # new instance every time
    df1[i] = le.fit_transform(df1[i])
    encoders[i] = le # storing the correct encoder for each column
    mappings[i] = dict(zip(le.classes_,le.transform(le.classes_)))
```

```python
encoders
```

```python
mappings
```

```python
df1.info()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error,mean_absolute_percentage_error
```

```python
X = df1.drop(columns = ['selling_price']) # all independent cols
y = df1['selling_price'] # dependent cols
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size = 0.80,random_state = 42)
```

```python
my_model = LinearRegression()
```

```python
my_model.fit(X_train,y_train)
```

```python
y_pred = my_model.predict(X_test)
```

```python
r2_score(y_test,y_pred)
```
