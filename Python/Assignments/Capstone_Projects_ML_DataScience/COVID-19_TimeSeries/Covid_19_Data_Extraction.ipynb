```python
# Grouped by day, country
# =======================

full_grouped = full_table.groupby(['Date', 'Country/Region'])[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum().reset_index()
```

```python
full_grouped
```

```python
full_grouped.groupby(['Country/Region', 'Date', ])[['Confirmed', 'Deaths', 'Recovered']].sum()
```

```python
temp = full_grouped.groupby(['Country/Region', 'Date', ])[['Confirmed', 'Deaths', 'Recovered']]
temp = temp.sum().diff().reset_index() # diff calculates difference between this row and previous row
```

```python
temp
```

```python
temp['Country/Region'].shift(1)
```

```python
temp['Country/Region'] != temp['Country/Region'].shift(1)

# Move the entire Country/Region column down by 1 row, and fill the first row with NaN.
# Means it takes value of the previous row for that COLumn only
# that column's last row value cannot be seen in new dataframe
```

```python
mask = temp['Country/Region'] != temp['Country/Region'].shift(1)

# we are doing this
# because the 1st row of a country other than 1st country, assumes the previous day cases are last day of previous country
# which is wrong. the 1st row of any country has no previous day cases, so the New Cases would be Nan

temp.loc[mask, 'Confirmed'] = np.nan
temp.loc[mask, 'Deaths'] = np.nan
temp.loc[mask, 'Recovered'] = np.nan

# renaming columns
temp.columns = ['Country/Region', 'Date', 'New cases', 'New deaths', 'New recovered']
# =================================================================

# merging new values
full_grouped = pd.merge(full_grouped, temp, on=['Country/Region', 'Date'])

# filling na with 0
full_grouped = full_grouped.fillna(0)

# fixing data types
cols = ['New cases', 'New deaths', 'New recovered']
full_grouped[cols] = full_grouped[cols].astype('int')
```

```python
full_grouped['New cases'] = full_grouped['New cases'].apply(lambda x: 0 if x<0 else x)
full_grouped['New deaths'] = full_grouped['New deaths'].apply(lambda x: 0 if x<0 else x)
full_grouped['New recovered'] = full_grouped['New recovered'].apply(lambda x: 0 if x<0 else x)


full_grouped.head()
```

```python
who_region = (
    full_table[['Country/Region', 'WHO Region']]
    .drop_duplicates()
    .set_index('Country/Region')['WHO Region']
    .to_dict()
)
```

```python
# add 'WHO Region' column
full_grouped['WHO Region'] = full_grouped['Country/Region'].map(who_region)

# find missing values
full_grouped[full_grouped['WHO Region'].isna()]['Country/Region'].unique()
```

```python
# Check province ends with country
exceptions = {
    "Canada": "Manitoba",
    "China": "Gansu",
    "Australia": "South Australia"
}

def match_province(row):
    country = row['Country/Region']
    province = str(row['Province/State']).strip()

    # If country has an exception rule
    if country in exceptions:
        return province == exceptions[country]

    # Normal rule for all other countries
    return province.endswith(country)

mask = full_table.apply(match_province, axis=1)

matching_provinces = full_table[mask][['Province/State', 'Country/Region']].drop_duplicates()
matching_provinces
```

```python
lat_long = (
    full_table[mask][['Country/Region', 'Lat', 'Long']]
    .drop_duplicates()
)
```

```python
full_grouped[full_grouped['Confirmed']!=0] \
                                    .groupby('Date')['Country/Region'] \
                                    .unique() \
                                    .apply(len)\
                                    .values
```

```python
len(full_grouped[full_grouped['Confirmed']!=0] \
                                    .groupby('Date')['Country/Region'] \
                                    .unique() \
                                    .apply(len)\
                                    .values)
```

```python
# Day wise
# ========

# table
day_wise = full_grouped.groupby('Date')[['Confirmed', 'Deaths', 'Recovered',
                                        'Active', 'New cases', 'New deaths', 'New recovered']].sum().reset_index()

# number cases per 100 cases
day_wise['Deaths / 100 Cases'] = round((day_wise['Deaths']/day_wise['Confirmed'])*100, 2)
day_wise['Recovered / 100 Cases'] = round((day_wise['Recovered']/day_wise['Confirmed'])*100, 2)
day_wise['Deaths / 100 Recovered'] = round((day_wise['Deaths']/day_wise['Recovered'])*100, 2)

# no. of countries
day_wise['No. of countries'] = full_grouped[full_grouped['Confirmed']!=0] \
                                    .groupby('Date')['Country/Region'] \
                                    .unique() \
                                    .apply(len)\
                                    .values

# fillna by 0
cols = ['Deaths / 100 Cases', 'Recovered / 100 Cases', 'Deaths / 100 Recovered']
day_wise[cols] = day_wise[cols].fillna(0)

day_wise.head()
```

```python
# save as .csv file
day_wise.to_csv('day_wise.csv', index=False)
```

```python
# Country wise
# ============

full_grouped['Date'] = pd.to_datetime(full_grouped['Date'])

# getting latest values, we get 1 row per country
country_wise = full_grouped[full_grouped['Date']==max(full_grouped['Date'])] \
                    .reset_index(drop=True) \
                    .drop('Date', axis=1)  # date is same for all selected rows

print(country_wise.shape)


# per 100 cases
country_wise['Deaths / 100 Cases'] = round((country_wise['Deaths']/country_wise['Confirmed'])*100, 2)
country_wise['Recovered / 100 Cases'] = round((country_wise['Recovered']/country_wise['Confirmed'])*100, 2)
country_wise['Deaths / 100 Recovered'] = round((country_wise['Deaths']/country_wise['Recovered'])*100, 2)

cols = ['Deaths / 100 Cases', 'Recovered / 100 Cases', 'Deaths / 100 Recovered']
country_wise[cols] = country_wise[cols].fillna(0)


# 1 week increase and % change
# ============================

today = full_grouped[full_grouped['Date']==max(full_grouped['Date'])] \
            .reset_index(drop=True) \
            .drop('Date', axis=1)[['Country/Region', 'Confirmed']]

last_week = full_grouped[full_grouped['Date']==max(full_grouped['Date'])-timedelta(days=7)] \
                .reset_index(drop=True) \
                .drop('Date', axis=1)[['Country/Region', 'Confirmed']]

temp = pd.merge(today, last_week, on='Country/Region', suffixes=(' today', ' last week')) # suffix since column from both df to be merged has same name
temp['1 week change'] = temp['Confirmed today'] - temp['Confirmed last week']
temp = temp[['Country/Region', 'Confirmed last week', '1 week change']]

country_wise = pd.merge(country_wise, temp, on='Country/Region')
country_wise['1 week % increase'] = round(country_wise['1 week change']/country_wise['Confirmed last week']*100, 2)

# to bring WHO to last, pop() removes the column and returns it, and assigning it back places it at the end.
country_wise['WHO Region'] = country_wise.pop('WHO Region')
country_wise.head()
```
