```python
def fill_null_provinces(df):
    """
    Fills null values in 'Province/State' column based on country-specific logic:
    - If all provinces for a country are null → fill with country name
    - If some provinces are null and others are not → fill with 'Mainland <CountryName>'
    - If province equals country name → leave as-is
    """
    df1 = df.copy()

    # Group by country
    for country in df1['Country/Region'].unique():
        country_mask = df1['Country/Region'] == country
        province_values = df1.loc[country_mask, 'Province/State']

        # Case 1: All provinces null
        if province_values.isnull().all():
            df1.loc[country_mask, 'Province/State'] = country

        # Case 2: Some provinces null, some not
        elif province_values.isnull().any():
            df1.loc[(country_mask) & (df['Province/State'].isnull()), 'Province/State'] = f'Mainland {country}'

        # Case 3: Province equals country name → leave as-is (already handled implicitly)

    return df1
```

```python
def check_date_continuity(g):
    g = g.sort_values('Date')
    all_dates = pd.date_range(g['Date'].min(), g['Date'].max(), freq='D')
    missing = all_dates.difference(g['Date'])
    return pd.Series({
        'start_date': g['Date'].min(),
        'end_date': g['Date'].max(),
        'expected_days': len(all_dates),
        'actual_days': g['Date'].nunique(),
        'missing_days': len(missing),
        'missing_dates': list(missing)
    })

result_dates = (
    df_cleaned.groupby(['Country/Region', 'Province/State'])
      .apply(check_date_continuity, include_groups=True)
      .reset_index()
)
```
